````python
# 1. Get user input using input(“Enter your age: ”). If user is 18 or older, give feedback: You are old enough to drive. If below 18 give feedback to wait for the missing amount of years. Output:
age = int(input("Enter your age: "))

if age >= 18:
    print("You are old enough to drive")
else:
    age_missing = 18 - age
    print(f"You need {age_missing} more years to learn to drive")
````

##### 3) Get two numbers from the user using input prompt. 

. If a is greater than b return a is greater than b, 

. if a is less b return a is smaller than b, else a is equal to b. 

Expected Output:

Enter number one: 4

Enter number two: 3

4 is greater than 3

````python
a = int(input("Value for a: "))
b = int(input("Value for b: "))
if a > b:
    print(f"{a} is greater than {b}")
elif a < b:
    print(f"{a} is smaller than {b}")
else:
    print(f"{a} and {b} are equal")
````

##### 4) Write a code which gives grade to students according to theirs scores:
```sh
90-100, A
80-89, B
70-79, C
60-69, D
0-59, F
```

````python
grade = int(input("Your grade is: "))
if grade >= 0 and grade <= 59:
    print("F")
elif grade >= 60 and grade <= 69:
    print("D")
elif grade >= 70 and grade <= 79:
    print("C")
elif grade >= 80 and grade <= 89:
    print("B")
elif grade >= 90 or grade <= 100:
    print("A")
else:
    print("You didn't even attend the exam!")
````

##### 6) If a fruit doesn't exist in the list add the fruit to the list and print the modified list. If the fruit exists print('That fruit already exist in the list')

fruits = ['banana', 'orange', 'mango', 'lemon']

````python
fruits = [
    "Apple", "Banana", "Orange", "Mango", "Strawberries",
    "Blueberries", "Pineapple", "Kiwi", "Grapes", "Watermelon",
    "Peach", "Cherry", "Plum", "Pomegranate", "Guava",
    "Lemon", "Lime", "Grapefruit", "Dragon fruit", "Avocado"
]

fruit = input("Enter a fruit: ")

if fruit in fruits:
    print(f"That {fruit} already exist in the list")
else:
    fruits.append(fruit)
    print(fruits)
````
